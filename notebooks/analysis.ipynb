```python
import pandas as pd 
df = pd.read_csv("../data/togo-dapaong_qc.csv")
```

```python
import seaborn as sns 
import matplotlib.pyplot as plt 
sns.pairplot(df[["GHI", "DNI", "DHI", "Tamb"]].sample(1000))
plt.show()
```

```python
sns.boxplot(data=df[["GHI", "DNI"]]) 
plt.title("Boxplot of GHI and DNI") 
plt.show() 
```

```python
sns.histplot(df["GHI"], kde=True, bins=20, color="blue") 
plt.title("Distribution of GHI") 
plt.xlabel("GHI") 
plt.ylabel("Frequency") 
plt.show()  
```

```python
sns.countplot(data=df, x='DNI', color='steelblue') 
plt.title("count of Each DNI") 
plt.xlabel("DNI") 
plt.ylabel("count") 
plt.show() 
```
